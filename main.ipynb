#### 編碼處理規則
- 97 / 997 (其他)：
這是有效的類別選項，保留即可。但在做回歸分析時，通常會把「其他」歸類到參照組或樣本數較少的那一組。
- -10/ 5 (跳答、不適用)：
含義： 該問題不針對此人。例如：「請問您配偶的年齡？」若受訪者未婚，此題會是 -10。
動作：
若你研究的是「全體樣本」： 需小心處理。例如你想算「全體樣本的配偶平均年齡」，這在邏輯上不通，應該先__篩選__出「有配偶的人」。
若題目是條件式的： 例如「每週加班時數」，__沒工作的人是 -10。你應該把它改成 0__（代表沒加班），而不是刪除，否則你會高估平均加班量。
總結建議： 檢查題目邏輯。如果是「沒有這項特徵（如沒配偶）」，轉為 NaN 並在分析時排除這些人；如果是「量值為零（如沒加班）」，轉為 0。
- -8 (拒答) / -6 (不知道、不清楚)：
含義： 這是__受訪者的心理狀態，這本身就是一種數據__。
動作：
__連續變數（薪資）__： 轉為 NaN。你無法用 -8 去算平均薪資。
__類別變數（政治傾向、滿意度）__： 可以保留。有時候「拒答」的人可能具有某種特殊特質（例如高收入者常拒答薪資）。若你發現 -8 比例很高（如 >5%），建議在敘述統計中特別說明；若比例很低，轉為 NaN 即可。
- -2 (虧損或打平)：
動作： 這是針對金額題。如果題目是__「獲利」__，建議改為 0 或負數（如果問卷有提供虧損金額）。如果是問__「收入」__，打平通常意味著淨收入為 0。建議改為 0。
- -1 (數值不固定)：
動作： 例如「每月給父母多少錢？」，回答「不固定」。這在統計上很難處理。建議轉為 NaN，除非你要把它變成一個 __類別變數（固定給錢 vs. 不固定 vs. 不給）__。
---
- -11 (不合理值)： 這是輸入錯誤或邏輯錯誤。
動作： 直接轉為 NaN。
- -9 (遺漏值)： 受訪者漏填。
動作： 直接轉為 NaN。
- -5 (數值超過上限)： 例如薪水太高超過欄位位數。
動作： 轉為 NaN，或者（進階做法）替換為該欄位的__最大值__（Winsorization），但初學者建議轉 NaN 較保險。

解決方案：純 R 手作 KNN 補值 (無痛版)

```r
# ==========================================
# 1. 載入基本工具 (跳過 VIM)
# ==========================================
if (!require("dplyr")) install.packages("dplyr")
library(dplyr)

# 複製一份資料
df_work <- independent_child_var

# ==========================================
# 2. 設定變數與預處理 (與 Python 完全一致)
# ==========================================
target_col <- 'e04z05'
feature_cols <- c('a02z01', 'a01', 'b03')
cols_to_process <- c(target_col, feature_cols)

# 轉換數值
df_work[cols_to_process] <- lapply(df_work[cols_to_process], function(x) as.numeric(as.character(x)))

# A. 目標欄位：負數轉 NA
df_work[[target_col]][df_work[[target_col]] < 0] <- NA

# B. 輔助欄位：負數轉 NA 並用中位數補齊 (KNN 特徵不能有空值)
for (col in feature_cols) {
  df_work[[col]][df_work[[col]] < 0] <- NA
  m_val <- median(df_work[[col]], na.rm = TRUE)
  df_work[[col]][is.na(df_work[[col]])] <- m_val
}
# ==========================================
# 3. 標準化 (精度校準版)
# ==========================================
# 確保與 Python MinMaxScaler 行為完全一致
scale_min_max <- function(x) {
  # 排除所有 NA 計算真正的 min/max
  rng <- range(x, na.rm = TRUE)
  if (rng[2] == rng[1]) return(x * 0)
  return((x - rng[1]) / (rng[2] - rng[1]))
}

df_scaled <- df_work %>%
  select(all_of(cols_to_process)) %>%
  mutate(across(everything(), scale_min_max))

# ==========================================
# 4. 執行 KNN 填補 (穩定排序版)
# ==========================================
na_idx <- which(is.na(df_scaled[[target_col]]))
complete_idx <- which(!is.na(df_scaled[[target_col]]))

ref_data <- df_scaled[complete_idx, ]
query_data <- df_scaled[na_idx, ]

cat("正在校準補值...\n")

imputed_values_scaled <- sapply(1:nrow(query_data), function(i) {
  # 計算歐幾里德距離
  diffs <- sweep(ref_data[, feature_cols], 2, as.numeric(query_data[i, feature_cols]))
  dist_sq <- rowSums(diffs^2)

  # 精度校準：對齊浮點數誤差
  dist_sq <- round(dist_sq, 12)

  # 取得前 5 個最近鄰居
  # 使用穩定排序，若距離相等則按原始索引排序
  nearest_neighbors <- order(dist_sq, complete_idx)[1:5]

  # 計算平均值
  return(mean(ref_data[[target_col]][nearest_neighbors]))
})

# ==========================================
# 5. 還原與後處理
# ==========================================
t_rng <- range(df_work[[target_col]], na.rm = TRUE)
filled_values_raw <- (imputed_values_scaled * (t_rng[2] - t_rng[1])) + t_rng[1]

# 關鍵點：使用與 Python 相同的四捨五入邏輯
# R 的 round() 預設是 "round to even" (2.5 變 2, 3.5 變 4)
# 如果 Python 那邊是 2.5 變 3，則需要手動調整
custom_round <- function(x) { floor(x + 0.5) } # 傳統四捨五入

# 嘗試切換不同的 round 方式看是否能對齊那 3 筆
filled_values <- round(filled_values_raw)
filled_values <- pmin(pmax(filled_values, 1), 5)

# 填回
independent_child_var$e04z05_filled <- independent_child_var[[target_col]]
# 處理原本的負數/NA
missing_mask <- (independent_child_var[[target_col]] < 0 | is.na(independent_child_var[[target_col]]))
independent_child_var$e04z05_filled[missing_mask] <- filled_values
independent_child_var$e04z05_filled <- as.integer(independent_child_var$e04z05_filled)

# ==========================================
# 6. 檢查結果
# ==========================================
print(table(independent_child_var$e04z05_filled))
```

### 第一部分：為什麼用 KNN 補「收入」？（方法論支持）

收入（Income）資料通常具有 **右偏（Right-skewed）** 特性（少數人極高薪，多數人一般），且與個人背景高度相關。傳統的「平均數填補法」會低估變異數且受極端值影響，而 KNN 能保留資料的局部結構。

*   **引用文獻 1：KNN 優於平均數填補**
    *   **Batista, G. E., & Monard, M. C. (2003).** "An analysis of four missing data treatment methods for supervised learning." *Applied Artificial Intelligence*, 17(5), 519-533.
    *   **論點：** 該研究比較了多種補值方法，證實 KNN 在處理多變量資料缺失時，表現優於傳統的平均數填補（Mean Imputation），因為它利用了樣本間的相似性（Similarity），更能還原真實的數據分佈。

*   **引用文獻 2：KNN 適用於社經數據**
    *   **Vatsa, M., Singh, R., & Noore, A. (2005).** "Imputation of missing data using adaptive k-nearest neighbours."
    *   **論點：** 指出在包含數值與類別混合的資料集（如人口統計變數）中，KNN 能夠有效利用輔助變數（如教育、職業）來預測目標變數（如收入）。

### 第二部分：為什麼選「性別、年齡、教育、工作」當特徵？（理論支持）

這四個變數不是隨便選的，它們是經典 **「明瑟薪資方程式（Mincer Earnings Function）」** 的核心組成部分。這證明了你的特徵選擇具有深厚的經濟學理論基礎。

#### 1. 教育程度 (Education) & 年齡 (Age)
這是「人力資本（Human Capital）」的核心。收入取決於你的技能（教育）和經驗（年齡/年資）。

*   **經典文獻 (必引)：**
    *   **Mincer, J. (1974).** *Schooling, Experience, and Earnings*. New York: Columbia University Press.
    *   **論點：** 著名的明瑟方程式：$\ln(Income) = \alpha + \beta_1(Education) + \beta_2(Experience) + \beta_3(Experience)^2$。該理論證實了教育年數與工作經驗（通常以**年齡**為代理變數）是解釋個人薪資差異最強的兩個預測因子。

*   **Becker, G. S. (1964).** *Human Capital: A Theoretical and Empirical Analysis*.

### 第四部分：如何在報告中撰寫？（範例）

你可以將上述文獻整合成一段強而有力的 **「資料處理與補值策略」** 說明：

> **3.2 遺漏值處理 (Missing Data Imputation)**
>
> 在父母樣本的年收入變數（k22/k23）中，存在 8 筆（約 7.2%）的隨機遺漏值。為避免刪除樣本導致統計檢定力降低，本研究採用 **KNN（K-Nearest Neighbors）演算法** 進行數值填補。相較於平均數填補，KNN 能保留資料的局部結構與變異性，更適用於收入這類非常態分佈的資料（Batista & Monard, 2003）。
>
> 在 KNN 演算法的特徵變數（Features）選擇上，本研究依據 **人力資本理論（Human Capital Theory）** 與 **明瑟薪資方程式（Mincer Earnings Function）**（Mincer, 1974; Becker, 1964），選取 **「教育程度」** 與 **「年齡」** 作為預測收入的核心指標。此外，考量勞動市場的結構性差異（Blau & Kahn, 2017），本研究亦納入 **「性別」** 與 **「工作狀態」** 作為輔助特徵，以確保填補值能反映受訪者真實的社經背景特徵。

這段話把你的 **方法（KNN）** 和 **特徵選擇（人力資本+性別）** 全部用文獻鎖死，審查者或讀者看到這裡就會知道你是內行的。

```r
# ==========================================
# 1. 環境準備與資料清洗
# ==========================================
if (!require("dplyr")) install.packages("dplyr")
library(dplyr)

# 建立工作副本
df_knn <- independent_parent_var

target_col <- 'k22'
feature_cols <- c('c02r1', 'a01', 'a02z01', 'b01')
cols_to_process <- c(target_col, feature_cols)

# 確保所有欄位轉為數值 (Numeric)
df_knn[cols_to_process] <- lapply(df_knn[cols_to_process], function(x) as.numeric(as.character(x)))

# A. 輔助特徵清洗：處理負值並用中位數補齊 (KNN 特徵不可有 NA)
for (col in feature_cols) {
  df_knn[[col]][df_knn[[col]] < 0] <- NA
  m_val <- median(df_knn[[col]], na.rm = TRUE)
  df_knn[[col]][is.na(df_knn[[col]])] <- m_val
}

# B. 目標變數清洗：將 k22 小於 0 的值轉為 NA (這是我們要補的洞)
df_knn[[target_col]][df_knn[[target_col]] < 0] <- NA

cat(sprintf("補值前 %s 的遺漏數量: %d\n", target_col, sum(is.na(df_knn[[target_col]]))))

# ==========================================
# 2. 標準化 (Min-Max Scaling)
# ==========================================
# 手動實作 MinMaxScaler，確保年齡與收入權重合理
df_scaled <- df_knn %>% select(all_of(cols_to_process))

# 紀錄目標欄位的 Min/Max 以便稍後還原
target_min <- min(df_scaled[[target_col]], na.rm = TRUE)
target_max <- max(df_scaled[[target_col]], na.rm = TRUE)

for(col in cols_to_process) {
  c_min <- min(df_scaled[[col]], na.rm = TRUE)
  c_max <- max(df_scaled[[col]], na.rm = TRUE)
  if(c_max - c_min != 0) {
    df_scaled[[col]] <- (df_scaled[[col]] - c_min) / (c_max - c_min)
  }
}

# ==========================================
# 3. 執行 KNN 補值 (k=5)
# ==========================================
na_idx <- which(is.na(df_scaled[[target_col]])) # 遺漏值索引
comp_idx <- which(!is.na(df_scaled[[target_col]])) # 完整資料索引

if(length(na_idx) > 0) {
  cat("正在進行 KNN 運算，處理", length(na_idx), "筆遺漏資料...\n")

  ref_data <- df_scaled[comp_idx, ]
  query_data <- df_scaled[na_idx, ]

  # 使用向量運算計算歐幾里德距離
  imputed_scaled <- sapply(1:nrow(query_data), function(i) {
    # 計算距離: sqrt(sum((p-q)^2)) -> 這裡省略 sqrt 不影響排序
    dist_sq <- rowSums(sweep(ref_data[, feature_cols], 2, as.numeric(query_data[i, feature_cols]))^2)
    # 找出最近的 5 個鄰居
    neighbors <- order(dist_sq)[1:5]
    # 取鄰居目標值的平均
    return(mean(ref_data[[target_col]][neighbors]))
  })

  # 還原數值 (Inverse Transform)
  filled_values <- (imputed_scaled * (target_max - target_min)) + target_min
} else {
  cat("沒有需要補值的空缺。\n")
}

# ==========================================
# 4. 後處理與回填
# ==========================================
# 四捨五入並限制不小於 0 (Clip lower=0)
filled_values <- pmax(round(filled_values), 0)

# 建立新欄位 k22_filled 並回填
independent_parent_var$k22_filled <- independent_parent_var[[target_col]]
# 處理原本的負數
independent_parent_var$k22_filled[independent_parent_var$k22_filled < 0] <- NA
# 填補 NA 位置
independent_parent_var$k22_filled[na_idx] <- filled_values
# 轉為整數
independent_parent_var$k22_filled <- as.integer(independent_parent_var$k22_filled)

# ==========================================
# 5. 驗證與結果檢視
# ==========================================
cat(rep("-", 30), "\n")
cat(sprintf("補值後 k22_filled 的遺漏數量: %d\n", sum(is.na(independent_parent_var$k22_filled))))

# 檢視特定案例 (轉換為 R 的 1-based index)
problem_indices <- c(393, 462, 793, 1134, 1462, 2093, 2301, 3470, 3474, 3761, 4198, 4449)
# 這裡假設你的 rownames 跟原本的 index 是一致的
existing_cases <- independent_parent_var[rownames(independent_parent_var) %in% as.character(problem_indices), ]

if(nrow(existing_cases) > 0) {
  cat("\n特定案例補值結果檢視:\n")
  print(existing_cases[, c('c02r1', 'a01', 'a02z01', 'b01', 'k22', 'k22_filled')])
}
```

### 單變量檢定

```r
library(dplyr)

# 確保 target 是有序變數（1~4），並處理缺失值
target_col <- 'd07'
df <- merged_df %>% dplyr::select(all_of(c(target_col, numeric, ordinal, binary)))

# 移除 target 缺失的列
df <- df %>% tidyr::drop_na(all_of(target_col))

# 初始化結果列表
results <- list()

# 1. Numerical variables → Kruskal-Wallis H-test
# 將 numerical 變數依 target 分組（因 target 是有序類別 1~4，視為分組變數）
for (var in numeric) {
  # 移除該變數的缺失值
  temp_df <- df %>% dplyr::select(all_of(c(target_col, var))) %>% tidyr::drop_na()
  if (nrow(temp_df) == 0) {
    next
  }

  groups <- split(temp_df[[var]], temp_df[[target_col]])
  # 至少需要兩組且每組有資料
  if (length(groups) < 2 || any(sapply(groups, length) == 0)) {
    next
  }

  # Kruskal-Wallis 檢定
  # formula: dependent_variable ~ grouping_variable
  kruskal_result <- kruskal.test(reformulate(target_col, response = var), data = temp_df)

  results <- append(results, list(data.frame(
    Variable = var,
    Type = 'Numerical',
    Test = 'Kruskal-Wallis',
    Statistic = kruskal_result$statistic,
    p.value = kruskal_result$p.value
  )))
}

# 2. Binary variables → Mann-Whitney U test
for (var in binary) {
  temp_df <- df %>% dplyr::select(all_of(c(target_col, var))) %>% tidyr::drop_na()
  if (nrow(temp_df) == 0) {
    next
  }

  # 確保 binary 變數只有 0/1 或類似二元值
  unique_vals <- sort(unique(temp_df[[var]]))
  if (length(unique_vals) != 2) {
    # 若非標準二元，可考慮跳過或處理
    next
  }

  # Mann-Whitney U 檢定 (wilcox.test 是 Wilcoxon rank sum test，等同於 Mann-Whitney U)
  # formula: dependent_variable ~ grouping_variable
  wilcox_result <- tryCatch({
    wilcox.test(reformulate(var, response = target_col), data = temp_df, exact = FALSE, alternative = 'two.sided')
  }, error = function(e) {
    # Handle cases where ranks might be identical causing errors
    NULL
  })

  if (!is.null(wilcox_result)) {
    results <- append(results, list(data.frame(
      Variable = var,
      Type = 'Binary',
      Test = 'Mann-Whitney U',
      Statistic = wilcox_result$statistic,
      p.value = wilcox_result$p.value
    )))
  }
}

# 3. Ordinal variables → Spearman rank correlation
for (var in ordinal) {
  temp_df <- df %>% dplyr::select(all_of(c(target_col, var))) %>% tidyr::drop_na()
  if (nrow(temp_df) == 0) {
    next
  }

  # Spearman 相關 (處理缺失後)
  cor_result <- cor.test(x = temp_df[[target_col]], y = temp_df[[var]], method = 'spearman',exact = FALSE)

  results <- append(results, list(data.frame(
    Variable = var,
    Type = 'Ordinal',
    Test = 'Spearman',
    Statistic = cor_result$estimate,
    p.value = cor_result$p.value
  )))
}

# 轉為 DataFrame 並顯示
results_df <- do.call(rbind, results)
print(results_df)

# 可選：保存結果
# write.csv(results_df, 'univariate_tests_results.csv', row.names = FALSE)
```

### VIF

```r
library(dplyr)

# 1. 準備資料
all_predictors <- c(numeric, ordinal, binary)
X <- merged_df %>% dplyr::select(all_of(all_predictors))

# 強制轉為數值型態
X_numeric <- as.data.frame(lapply(X, function(x) as.numeric(as.character(x))))

# 移除含有 NA 的列
X_clean <- na.omit(X_numeric)

cat(sprintf("原始樣本數: %d\n", nrow(merged_df)))
cat(sprintf("VIF 分析使用樣本數 (無缺失): %d\n", nrow(X_clean)))
cat(sprintf("變數數量: %d\n", ncol(X_clean)))

# ==========================================
# 2. 自定義 VIF 計算函數 (不需依賴 car 套件)
# ==========================================
calculate_vif <- function(df) {
  vars <- colnames(df)
  vif_values <- numeric(length(vars))
  names(vif_values) <- vars

  for (i in seq_along(vars)) {
    # 輪流將每個變數當作應變數 (y)，其餘當作自變數 (x)
    form <- as.formula(paste(vars[i], "~ ."))
    # 建立線性迴歸模型
    fit <- lm(form, data = df)
    # 提取 R-squared
    r_sq <- summary(fit)$r.squared
    # 計算 VIF = 1 / (1 - R^2)
    # 若 R^2 為 1，VIF 會變成無限大 (Inf)
    vif_values[i] <- 1 / (1 - r_sq)
  }
  return(vif_values)
}

# ==========================================
# 3. 執行計算與顯示
# ==========================================
if (ncol(X_clean) > 1) {
  cat("正在計算 VIF，請稍候...\n")

  # 呼叫自定義函數
  vif_res <- calculate_vif(X_clean)

  # 轉為 dataframe 並排序
  vif_data <- data.frame(
    Variable = names(vif_res),
    VIF = as.numeric(vif_res)
  ) %>%
    arrange(desc(VIF))

  cat("\n【VIF 結果（高 → 低）】\n")
  print(vif_data)

  # 提示高共線性變數 (通常 VIF > 5 或 10)
  high_vif <- vif_data %>% filter(VIF > 10)
  if(nrow(high_vif) > 0) {
    cat("\n警告：以下變數 VIF > 10，存在嚴重共線性問題：\n")
    print(high_vif$Variable)
  }

} else {
  cat("\n變數數量不足，無法計算 VIF。\n")
}
```

```r
library(dplyr) # 載入 dplyr 函式庫以進行資料操作 (若尚未載入)

# 從 Model 2 的摘要中提取係數表
coef_table_from_summary <- as.data.frame(summary(mod2)$coefficients)

# 計算 p 值 (假設 't value' 在這裡可視為 Z 值)
coef_table_from_summary$p_value <- 2 * pnorm(-abs(coef_table_from_summary$`t value`))

# 將變數名稱從 rownames 轉換為一個欄位
results_df_r <- coef_table_from_summary %>%
  tibble::rownames_to_column(var = "Variable")

# 計算 Odds Ratio
results_df_r$Odds_Ratio <- exp(results_df_r$Value)

# 過濾掉截距項 (通常是最後三行，在 R 的 polr 摘要中名稱為 '1|2', '2|3', '3|4')
num_rows <- nrow(results_df_r)
results_df_r_filtered <- results_df_r[1:(num_rows - 3), ]

# 選擇並重新命名欄位，以符合 Python 輸出的格式
final_output_df <- results_df_r_filtered %>%
  dplyr::select(
    Variable = Variable,
    coef = Value, # 'Value' 欄位在 R 中代表係數
    `P>|z|` = p_value, # 使用反引號以包含特殊字元
    Odds_Ratio = Odds_Ratio
  )

# 打印最終結果
print(final_output_df)
```

### 比例勝算假設（Proportional Odds Assumption）
很可能因為樣本數少不顯著，但還是跑跑看

```r
library(nnet) # For multinom function
library(MASS) # For polr function, already loaded in previous cell

# 1. 建立多元羅吉特模型 (作為對照組)
# 注意：這裡不需手動加常數項，因為 MNLogit 預設會處理，
# 但為了與 OrderedModel 比較，結構需一致。

# 為了不改變 `analysis_df$d07` 的有序因子屬性，複製一份資料用於 multinomial 模型的擬合
analysis_df_mn <- analysis_df
analysis_df_mn$d07 <- factor(analysis_df_mn$d07, ordered = FALSE)

# 建立模型公式，`multinom` 會自動加入截距項
formula_mn <- as.formula(paste("d07 ~", paste(final_model_vars, collapse = " + ")))

# 運行多元 Logit 模型 (method='bfgs', disp=False)
mod_mn <- multinom(formula_mn, data = analysis_df_mn, Hess = TRUE, trace = FALSE)

# 2. 提取兩個模型的對數概似值 (Log-Likelihood)
ll_ordered <- logLik(mod2) # mod2 是先前 fitting 的 OrderedModel
ll_multinomial <- logLik(mod_mn)

# 3. 計算似然比統計量 (LR statistic)
# LR = 2 * (LL_full - LL_restricted)
lr_stat <- 2 * (as.numeric(ll_multinomial) - as.numeric(ll_ordered))

# 4. 計算自由度差異 (df)
df_multinomial <- attr(ll_multinomial, "df")
df_ordered <- attr(ll_ordered, "df")
df_diff <- df_multinomial - df_ordered

# 5. 計算 P 值
p_value <- pchisq(lr_stat, df = df_diff, lower.tail = FALSE)

cat(sprintf("Likelihood Ratio Test 統計量: %.4f\n", lr_stat))
cat(sprintf("自由度差異: %.1f\n", df_diff))
cat(sprintf("P 值: %.4f\n", p_value))

if (p_value < 0.05) {
    cat("結果：拒絕原假設。違反比例勝算假設，建議改用 Generalized Ordered Model。\n")
} else {
    cat("結果：無法拒絕原假設。比例勝算假設成立，模型設定正確。\n")
}
```

### 簡化「父母婚姻解體」變數與注意交互作用項（催婚 x 父母婚姻解體）

```r
# 保留原始兩個婚姻解體變數
analysis_df$pressure_x_sep <- analysis_df$marriage_pressure_bin * analysis_df$h23z01
analysis_df$pressure_x_div <- analysis_df$marriage_pressure_bin * analysis_df$h23z02


control_vars <- c('a01', 'c02r1_child', 'g22z02c1', 'k22_filled')

# Model 3b: 主效應（3個） + 交互（2個） + 控制（4個） → 總共 9 個變數
interaction_model_vars <- c(control_vars,
    'marriage_pressure_bin',
    'h23z01',
    'h23z02',
    'pressure_x_sep',
    'pressure_x_div'
)

# Model 3: 包含交互作用
# R 的 polr 函數會自動處理 formula 中的變數選擇

# 確保 d07 是有序因子 (如果尚未處理)
if (!is.ordered(analysis_df$d07)) {
  analysis_df$d07 <- factor(analysis_df$d07, levels = c(1, 2, 3, 4), ordered = TRUE)
}

# 建立模型公式
formula3 <- as.formula(paste("d07 ~", paste(interaction_model_vars, collapse = " + ")))

# 運行有序 Logit 模型
mod3 <- polr(formula3, data = analysis_df, Hess = TRUE) # Hess=TRUE 為了計算標準誤

print("Model 3: Full Model with Interaction")
print(summary(mod3))
```

```r
# 比較 Model 2 與 Model 3
# Note: res2 and res3 are summary objects from polr, not directly containing llf and df_model
# We need to extract logLik and df from the original model objects (mod2 and mod3).

# Get log-likelihoods
ll_mod2 <- logLik(mod2)
ll_mod3 <- logLik(mod3)

# Calculate LR statistic
lr_stat <- 2 * (as.numeric(ll_mod3) - as.numeric(ll_mod2))

# Calculate degrees of freedom difference
df_mod2 <- attr(ll_mod2, "df")
df_mod3 <- attr(ll_mod3, "df")
df_diff <- df_mod3 - df_mod2

# Calculate P value
p_val <- pchisq(lr_stat, df = df_diff, lower.tail = FALSE)

cat(sprintf("Likelihood Ratio Test: \u03c7\u00b2(%.0f) = %.3f, p = %.3f\n", df_diff, lr_stat, p_val))
```
